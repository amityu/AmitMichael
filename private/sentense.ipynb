```python
import csv


def sentence_to_numbers(sentence):
    words = sentence.split(' ')
    groups = [words[i:i + 6] for i in range(0, len(words), 6)]

    for i, group in enumerate(groups):
        result_group = []
        for word in group:
            number = 0
            for letter in word:
                number = number * 26 + (ord(letter.lower()) - ord('a') + 1)
            result_group.append(number)
        with open(f'group_{i + 1}.csv', 'w', newline='') as file:
            writer = csv.writer(file)
            writer.writerow(result_group)
```

```python
sentence_to_numbers('a ab cdf a ab cdf a ab cdf a ghds fsa twer vzxcv tewrtew vzxcvz fgfasd twrtew fasdfas vzxcvz rwerwe fddas vczxv  vvcx rwerer')
```

```python
import pandas as pd
import matplotlib.pyplot as plt


def plot_groups(n_groups=4):
    for i in range(1, n_groups + 1):
        data = pd.read_csv(f'group_{i}.csv', header=None)
        plt.figure()
        plt.plot(data.loc[0], marker='o')
        plt.title(f'Group {i}')
        for j, value in enumerate(data.loc[0]):
            plt.text(j, value, str(value), ha='center')
    plt.show()


# Call the function
plot_groups()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# set style and font size
sns.set_style('darkgrid')
plt.rcParams.update({'font.size': 14})


def plot_groups(n_groups=4):
    for i in range(1, n_groups + 1):
        data = pd.read_csv(f'group_{i}.csv', header=None)
        plt.figure()
        plt.bar(range(len(data.loc[0])), data.loc[0], color='lightblue')
        plt.title(f'Group {i}')
        for j, value in enumerate(data.loc[0]):
            plt.text(j, value + 0.5, str(value), ha='center', color='darkred')
        plt.xticks(range(len(data.loc[0])))  # set x-ticks to be integers (not floats)
    plt.show()


# Call the function
plot_groups()
```
